What are the most demanded skills for the top 3 most popular data role?

```python
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt
import seaborn as sns

dataset=load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()

df['job_posted_date']=pd.to_datetime(df['job_posted_date'])
df['job_skills']=df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df_in=df[df['job_country']=='India']
df_skills=df_in.explode('job_skills')
```

```python
df_skills[['job_title','job_skills']]
df_skills_group=df_skills.groupby(['job_skills','job_title_short']).size()
```

```python
df_skills_group=df_skills_group.reset_index(name='skill_count')
df_skills_group.sort_values(by='skill_count', ascending=False, inplace=True)
df_skills_group
```

```python
job_titles=['Data Analyst', 'Data Scientist','Data Engineer']
job_titles.sort()
```

```python
df_job_title_count=df_in['job_title_short'].value_counts().reset_index(name='job_total')
```

```python
df_skill_perc=pd.merge(df_skills_group,df_job_title_count, how='left', on='job_title_short' )
df_skill_perc['skill_percent']=100*df_skill_perc['skill_count']/df_skill_perc['job_total']
df_skill_perc
```

```python
fig, ax=plt.subplots(len(job_titles),1)
sns.set_theme(style='ticks')
for i, job_title in enumerate(job_titles):
    df_plot=df_skill_perc[df_skill_perc['job_title_short']==job_title].head(5)
    #df_plot.plot(kind='barh' , x='job_skills', y='skill_percent', ax=ax[i], title=job_title)
    sns.barplot(data=df_plot, x='skill_percent', y='job_skills', ax=ax[i], hue='skill_count', palette='dark:y_r')
    ax[i].set_title(job_title)
    ax[i].set_ylabel('')
    ax[i].set_xlabel('')
    ax[i].legend().remove()
    ax[i].set_xlim(0,80)

    for n,v in enumerate(df_plot['skill_percent']):
        ax[i].text(v+1, n, f'{v:.0f}%', va='center')

    if i!=len(job_titles) - 1:
        ax[i].set_xticks([])

fig.suptitle('Counts of Top Skills in Job Postings', fontsize=15)
plt.tight_layout()
```
